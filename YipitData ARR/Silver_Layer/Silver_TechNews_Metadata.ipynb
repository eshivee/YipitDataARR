```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
metadata_df = spark.read.table('yipidata.silver.company_metadata')
```

```python
article_df = spark.read.table('yipidata.silver.company_articles')
```

```python
joined_df = article_df.join(metadata_df, article_df.company_name_new == metadata_df.company, 'left')
```

```python
unmatched_companies = joined_df.filter(col('company').isNull())
```

```python
final_df = joined_df.filter(col('company').isNotNull())
```

```python
final_df = final_df.drop('company_name_new')
```

```python
final_df.display()
```

```python
final_df = final_df.withColumn('company_age', col('year') - col('founded_year'))
```

```python
final_df.display()
```

```python
ai_related_companies = final_df.filter((col('category_new') == 'Artificial Intelligence') & (col('industry')   == 'AI/ML'))
```

```python
final_df.createOrReplaceTempView('source')
```

```python
%sql
CREATE TABLE IF NOT EXISTS yipidata.silver.articles_metadata_obt(
    article_id STRING,
    title STRING,
    summary STRING,
    url STRING,
    author STRING,
    word_count INT,
    category_new STRING,
    revenue DOUBLE,
    date_cleaned DATE,
    year INT,
    month STRING,
    quarter STRING,
    company STRING,
    employee_count BIGINT,
    founded_year BIGINT,
    headquarters STRING,
    industry STRING,
    is_public BOOLEAN,
    stock_ticker STRING,
    company_size STRING,
    company_age BIGINT
)
USING DELTA;
```

```python
%sql
MERGE INTO yipidata.silver.articles_metadata_obt AS trg
USING source AS src
ON trg.article_id = src.article_id
WHEN MATCHED THEN
UPDATE SET *
WHEN NOT MATCHED THEN
INSERT *
```

```python
ai_articles_enr = spark.sql("""SELECT * FROM yipidata.silver.articles_metadata_obt
                                WHERE (industry = 'AI/ML' OR category_new = 'Artificial Intelligence') 
                                AND (`year` BETWEEN 2022 AND 2024) 
                                AND (revenue > 50000000)""")
```

```python
ai_articles_enr.display()
```

```python
ai_articles_enr = ai_articles_enr.withColumn('embedding_text', concat_ws(' ', col('title'), col('summary')))
```

```python
ai_articles_pd = ai_articles_enr.toPandas()
```

```python
texts = ai_articles_pd['embedding_text'].tolist()
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings = model.encode(texts, show_progress_bar=True)
print(embeddings)
```

```python
ai_articles_pd['embeddings'] = [e.tolist() for e in embeddings]
```
