```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.readStream.format('cloudFiles')\
        .option('cloudFiles.format', 'csv')\
        .option('cloudFiles.schemaLocation', f'/Volumes/yipidata/bronze/checkpoint_location/tech_news')\
        .load(f"/Volumes/yipidata/source/source_data/tech_news/")
```

```python
df.writeStream.format('delta')\
        .outputMode('append')\
        .option('checkpointLocation', f'/Volumes/yipidata/bronze/checkpoint_location/tech_news')\
        .trigger(once = True)\
        .toTable(f'yipidata.bronze.tech_news')
```

```python
%sql
select * from yipidata.bronze.tech_news
```
