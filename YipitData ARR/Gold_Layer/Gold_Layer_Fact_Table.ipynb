```python
%sql
CREATE TABLE IF NOT EXISTS yipidata.gold.fact_table
AS 
SELECT SHA2(
        CONCAT(
            a.dim_article_id,
            '|',
            c.dim_company_id,
            '|',
            d.dim_date_id
        ),
        256
    ) AS arr_id,
 a.dim_article_id, d.dim_date_id, c.dim_company_id, f.revenue, f.company_age
FROM yipidata.silver.articles_metadata_obt f 
LEFT JOIN yipidata.gold.dim_article a
ON f.article_id = a.dim_article_id
LEFT JOIN yipidata.gold.dim_date d
ON f.date_cleaned = d.date_cleaned
LEFT JOIN yipidata.gold.dim_company c
ON f.company = c.company
```
