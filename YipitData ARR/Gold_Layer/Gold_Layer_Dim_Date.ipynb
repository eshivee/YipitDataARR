```python
%sql
CREATE TABLE IF NOT EXISTS yipidata.gold.dim_date
USING DELTA
AS

SELECT DISTINCT
    CAST(date_format(date_cleaned, 'yyyyMMdd') AS INT) AS dim_date_id,
    date_cleaned,
    year,
    month,
    quarter
FROM yipidata.silver.articles_metadata_obt
WHERE date_cleaned IS NOT NULL;
```
